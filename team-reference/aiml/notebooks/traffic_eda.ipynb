# Traffic Congestion EDA

This notebook profiles the provisional traffic dataset and identifies candidate congestion features.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/processed/processed_traffic.csv')
if not data_path.exists():
    data_path = Path('data/processed/processed_traffic.csv')
df = pd.read_csv(data_path, parse_dates=['timestamp'])
df.head()
```

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
display(df.isna().sum().rename('missing').to_frame())
display(df.describe(include='all'))
```

```python
df['density'] = df['vehicle_count'] / 200
df['is_peak_hour'] = df['hour'].between(8, 10) | df['hour'].between(17, 19)
print('Roads:', df['road_id'].nunique())
display(df.groupby('congestion_level')[['density', 'average_speed']].mean().sort_values('density'))
display(df.groupby('road_id')[['density', 'average_speed']].mean())
```

## Findings

Density, average speed, hour/peak status, and categorical road identity are candidate features. The sample covers only morning observations, so broader time-of-day conclusions require more data.
