```python
import pandas as pd

df = pd.read_csv("../data/traffic.csv")

df.head()
```

```python
print("Rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns)

print("\nMissing values:")
print(df.isnull().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# Convert DateTime from text to datetime
df["DateTime"] = pd.to_datetime(df["DateTime"])

# Rename columns according to TrafficVision schema
traffic = df.rename(columns={
    "ID": "id",
    "Junction": "road_id",
    "DateTime": "timestamp",
    "Vehicles": "vehicle_count"
})

# Keep only the required columns
traffic = traffic[
    ["id", "road_id", "timestamp", "vehicle_count"]
]

traffic.head()
```
